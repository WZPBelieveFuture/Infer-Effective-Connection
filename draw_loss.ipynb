# Stage2 Multiscale Loss Visualization

This notebook reads `train_loss0.csv` to `train_loss5.csv` and `test_loss0.csv` to `test_loss5.csv` from `loc_result_stage2/stage2_macro`, then plots the loss-vs-epoch curves for all six scales in a `2 x 3` grid.


```python
from __future__ import annotations
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from matplotlib.ticker import ScalarFormatter
```

```python
fig,ax1=plt.subplots(3,3, figsize= (8, 5))
scale_number = [16, 8, 2, 1]

left, right = 0, 0
train_loss = pd.read_csv('./loc_result_stage1/stage1_real_fmri_macro/train_scale_loss.csv',header=None)
train_loss = np.array(train_loss)
for i in range(train_loss.shape[1]):
    ax1[left][right].plot(np.arange(train_loss.shape[0])*50, train_loss[:,i], label='train_loss')
    ax1[left][right].set_title('scale=%s'%(scale_number[i]), fontsize=8)
    ax1[left][right].set_xlabel('epoch', fontsize=8)
    ax1[left][right].set_ylabel(r'loss', fontsize=8)
    ax1[left][right].xaxis.set_major_formatter(ScalarFormatter(useMathText=True))
    ax1[left][right].ticklabel_format(axis='x', style='sci', scilimits=(0,0))
    # ax1[left][right].xaxis.set_ticks(x + bar_width, freq_list)
    ax1[left][right].tick_params(labelsize=6)
    # if left ==0 and right ==0:
    ax1[left][right].legend(fontsize=5)
    right += 1   
    if right %3 == 0:     
        left += 1       
        right = 0   
    plt.subplots_adjust(wspace=0.4, hspace=0.9)  
```

```python
fig,ax1=plt.subplots(3,3, figsize= (8, 5))
scale_number = [32, 16, 8, 4, 2, 1]

ei_stage1, ei_stage2, ei_stage3 = [], [], []
left, right = 0, 0
train_loss = pd.read_csv('./loc_result_stage1/stage1_kuramoto_macro/train_loss.csv',header=None)
test_loss = pd.read_csv('./loc_result_stage1/stage1_kuramoto_macro/test_loss.csv',header=None)
train_loss, test_loss = np.array(train_loss), np.array(test_loss)
for i in range(train_loss.shape[1]):
    ax1[left][right].plot(np.arange(train_loss.shape[0])*50, train_loss[:,i], label='train_loss')
    ax1[left][right].plot(np.arange(train_loss.shape[0])*50,test_loss[:,i], label='test_loss')
    ax1[left][right].set_title('scale=%s'%(scale_number[i]), fontsize=8)
    ax1[left][right].set_xlabel('epoch', fontsize=8)
    ax1[left][right].set_ylabel(r'loss', fontsize=8)
    ax1[left][right].xaxis.set_major_formatter(ScalarFormatter(useMathText=True))
    ax1[left][right].ticklabel_format(axis='x', style='sci', scilimits=(0,0))
    # ax1[left][right].xaxis.set_ticks(x + bar_width, freq_list)
    ax1[left][right].tick_params(labelsize=6)
    # if left ==0 and right ==0:
    ax1[left][right].legend(fontsize=5)
    right += 1   
    if right %3 == 0:     
        left += 1       
        right = 0   
    plt.subplots_adjust(wspace=0.4, hspace=0.9)  
```

```python
fig,ax1=plt.subplots(3,3, figsize= (8, 5)) 
scale_number = [16, 8, 2, 1]
left, right = 0, 0
for weights in range(1, 5): 
    loss1 = pd.read_csv('./loc_result_stage2/stage2_kuramoto_macro/train_loss%s.csv'%(weights),header=None)
    loss2 = pd.read_csv('./loc_result_stage2/stage2_kuramoto_macro/test_loss%s.csv'%(weights),header=None)
    loss1, loss2 = np.array(loss1), np.array(loss2)
    ax1[left][right].plot(np.arange(loss1.shape[0])[:]*500, loss1[:], label='train loss')
    ax1[left][right].plot(np.arange(loss2.shape[0])[:]*500, loss2[:], label='test loss')
    ax1[left][right].xaxis.set_major_formatter(ScalarFormatter(useMathText=True))
    ax1[left][right].ticklabel_format(axis='x', style='sci', scilimits=(0,0))
    ax1[left][right].set_title('scale=%s'%(scale_number[weights-1]), fontsize=8)
    ax1[left][right].set_xlabel('epoch', fontsize=8)
    ax1[left][right].set_ylabel(r'loss', fontsize=8)
    # ax1[left][right].xaxis.set_ticks(x + bar_width, freq_list)
    ax1[left][right].tick_params(labelsize=6)
    if left ==0 and right ==0:
        ax1[left][right].legend(fontsize=5)
    right += 1 
    if right %3 == 0: 
        left += 1 
        right = 0  
    plt.subplots_adjust(wspace=0.5, hspace=0.9)
```
